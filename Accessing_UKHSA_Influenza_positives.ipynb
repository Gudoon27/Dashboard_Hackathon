```python
import requests
import time

class APIwrapper:

    _access_point = "https://api.ukhsa-dashboard.data.gov.uk"
    _last_access = 0.0

    def __init__(self, theme, sub_theme, topic,
                 geography_type, geography, metric):

        url_path = (
            f"/themes/{theme}/sub_themes/{sub_theme}/topics/{topic}"
            f"/geography_types/{geography_type}"
            f"/geographies/{geography}/metrics/{metric}"
        )

        self._start_url = APIwrapper._access_point + url_path

        self._filters = None
        self._page_size = -1
        self.count = None

    def get_page(self, filters=None, page_size=5):

        if filters is None:
            filters = {}

        if page_size > 365:
            raise ValueError("Max supported page size is 365")

        if filters != self._filters or page_size != self._page_size:

            self._filters = filters.copy()
            self._page_size = page_size
            self._next_url = self._start_url

        if self._next_url is None:
            return []

        curr_time = time.time()
        deltat = curr_time - APIwrapper._last_access

        if deltat < 0.33:
            time.sleep(0.33 - deltat)

        APIwrapper._last_access = time.time()

        parameters = {
            x: y for x, y in filters.items() if y is not None
        }

        parameters["page_size"] = page_size

        response = requests.get(
            self._next_url,
            params=parameters,
            timeout=30
        )

        response.raise_for_status()
        response = response.json()

        self._next_url = response["next"]
        self.count = response["count"]

        return response["results"]

    def get_all_pages(self, filters=None, page_size=365):

        data = []

        while True:

            next_page = self.get_page(filters, page_size)

            if next_page == []:
                break

            data.extend(next_page)

        return data
```

```python
combined = {}

for row in positivity:

    d = row["date"]

    combined.setdefault(
        d, {"date": d}
    )["positivity"] = row.get("metric_value")

influenza_timeseries = [
    combined[d] for d in sorted(combined)
]
```

```python
positivitydf = pd.read_pickle("positivitydf.pkl")
```

```python
year = wdg.Select(
    options=sorted(positivitydf.index.year.unique()),
    value=positivitydf.index.year[-1],
    rows=1,
    description="Year:",
    disabled=False
)
```

```python
def positivity_graph(graphyear):

    # Select data for chosen year
    yeardf = positivitydf[
        positivitydf.index.year == graphyear
    ]

    # Plot PCR positivity
    ax = yeardf["metric_value"].plot(
        figsize=(10, 5),
        color="blue",
        linewidth=2
    )

    ax.set_title(
        "Influenza PCR Positivity in England"
    )

    ax.set_xlabel("Date")
    ax.set_ylabel("PCR Positivity (%)")

    ax.grid(True)

    plt.show()


# Interactive output
output = wdg.interactive_output(
    positivity_graph,
    {"graphyear": year}
)

display(year, output)
```

```python
scale = wdg.RadioButtons(
    options=["linear", "log"],
    description="Scale:",
    disabled=False
)

controls = wdg.HBox([year, scale])


def positivity_timeseries_graph(graphyear, gscale):

    yeardf = positivitydf[
        positivitydf.index.year == graphyear
    ]

    logscale = gscale == "log"

    ax = yeardf["metric_value"].plot(
        figsize=(10, 5),
        logy=logscale
    )

    ax.set_title(
        "Influenza PCR Positivity in England"
    )

    ax.set_xlabel("Date")
    ax.set_ylabel("PCR Positivity (%)")

    plt.show()


graph = wdg.interactive_output(
    positivity_timeseries_graph,
    {
        "graphyear": year,
        "gscale": scale
    }
)

display(controls, graph)
```
